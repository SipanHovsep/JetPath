```python
import os

Left_dir = 'dataset/Left'
Right_dir = 'dataset/Right'
No_Sign_dir = 'dataset/No_Sign'

# we have this "try/except" statement because these next functions can throw an error if the directories exist already
try:
    os.makedirs(Right_dir)
    os.makedirs(Left_dir)
    os.makedirs(No_Sign_dir)  

except FileExistsError:
    print('Directories not created because they already exist')
```

```python
from jetbot import Robot
import time
import ipywidgets.widgets as widgets
from IPython.display import display
import traitlets


robot = Robot()

# create buttons
button_layout = widgets.Layout(width='100px', height='80px', align_self='center')
stop_button = widgets.Button(description='stop', button_style='danger', layout=button_layout)
forward_button = widgets.Button(description='forward', layout=button_layout)
backward_button = widgets.Button(description='backward', layout=button_layout)
left_button = widgets.Button(description='left', layout=button_layout)
right_button = widgets.Button(description='right', layout=button_layout)
no_sign_button = widgets.Button(description='add No Sign', button_style='info', layout=button_layout)  

# display buttons
middle_box = widgets.HBox([left_button, stop_button, right_button], layout=widgets.Layout(align_self='center'))
controls_box = widgets.VBox([forward_button, middle_box, backward_button])
display(controls_box)


def stop(change):
    robot.stop()
    
def step_forward(change):
    robot.forward(0.2)
    time.sleep(0.4)
    robot.stop()

def step_backward(change):
    robot.backward(0.2)
    time.sleep(0.5)
    robot.stop()

def step_left(change):
    robot.left(0.2)
    time.sleep(0.2)
    robot.stop()

def step_right(change):
    robot.right(0.2)
    time.sleep(0.2)
    robot.stop()
    
    
    # link buttons to actions
stop_button.on_click(stop)
forward_button.on_click(step_forward)
backward_button.on_click(step_backward)
left_button.on_click(step_left)
right_button.on_click(step_right)
```

```python
button_layout = widgets.Layout(width='128px', height='64px')
free_button = widgets.Button(description='add Right', button_style='success', layout=button_layout)
blocked_button = widgets.Button(description='add Left', button_style='danger', layout=button_layout)
no_sign_button = widgets.Button(description='add No Sign', button_style='info', layout=button_layout)  # <-- ADD THIS

free_count = widgets.IntText(layout=button_layout, value=len(os.listdir(Right_dir)))
blocked_count = widgets.IntText(layout=button_layout, value=len(os.listdir(Left_dir)))
no_sign_count = widgets.IntText(layout=button_layout, value=len(os.listdir(No_Sign_dir)))


display(widgets.HBox([free_count, free_button]))
display(widgets.HBox([blocked_count, blocked_button]))
display(widgets.HBox([no_sign_count, no_sign_button]))
```

```python
from uuid import uuid1

def save_snapshot(directory):
    image_path = os.path.join(directory, str(uuid1()) + '.jpg')
    with open(image_path, 'wb') as f:
        f.write(image.value)

def save_free():
    global Right_dir, free_count
    save_snapshot(Right_dir)
    free_count.value = len(os.listdir(Right_dir))
    
def save_blocked():
    global Left_dir, Left_count
    save_snapshot(Left_dir)
    blocked_count.value = len(os.listdir(Left_dir))
    
def save_no_sign():
    global No_Sign_dir, no_sign_count
    save_snapshot(No_Sign_dir)
    no_sign_count.value = len(os.listdir(No_Sign_dir))

    
# attach the callbacks, we use a 'lambda' function to ignore the
# parameter that the on_click event would provide to our function
# because we don't need it.
free_button.on_click(lambda x: save_free())
blocked_button.on_click(lambda x: save_blocked())
no_sign_button.on_click(lambda x: save_no_sign())
```
